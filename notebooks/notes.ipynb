# Windows 11 | AutoHotKey
#Requires AutoHotkey v2.0
```
; ==============================================================================
; 1. Caps Lock キーを完全に無効化
; ==============================================================================
*CapsLock::Return


; ==============================================================================
; 2. Escキーを押したとき、ノーマルモードに戻りつつIME（日本語入力）を確実にOFFにする
; ==============================================================================
~Esc::
{
    ; 現在アクティブなウィンドウのIME状態を強制的に「オフ（半角英語）」にする
    HWND := WinExist("A")
    IMC := DllCall("imm32\ImmGetDefaultIMEWnd", "Ptr", HWND, "Ptr")
    DllCall("SendMessage", "Ptr", IMC, "UInt", 0x0283, "UPtr", 0x006, "Ptr", 0)
    return
}

# Windows 11 | D:/Backup to Onedrive
```cmd
C:\Windows\System32>mklink /D "C:\Users\mrsmm\OneDrive\backup" "D:\backup"
symbolic link created for C:\Users\mrsmm\OneDrive\backup <<===>> D:\backup

C:\Windows\System32>
```
