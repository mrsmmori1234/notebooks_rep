```python
import os
import pandas as pd
import yfinance as yf
import deepl
from IPython.display import display, HTML

# 1. DeepLクライアントの初期化
api_key = os.environ.get("DEEPL_API_KEY")
if not api_key:
    raise ValueError("環境変数 'DEEPL_API_KEY' が設定されていません。")

translator = deepl.DeepLClient(api_key)

# 2. yfinanceを使って指定した銘柄や市場の最新ニュースを取得
# （例: シンガポールの代表的なETF [EWS] や、米国の主要指標 [^GSPC] などに変更可能です）
ticker_symbol = "EWS"  # iShares MSCI Singapore ETF (シンガポール株全体を追うETF)
print(f"{ticker_symbol} の最新関連ニュースを取得中...")

ticker = yf.Ticker(ticker_symbol)
news_items = ticker.news

if not news_items:
    print("ニュースが見つかりませんでした。")
else:
    # yfinanceのニュース構造からタイトルとリンクを抽出
    extracted_news = []
    for item in news_items[:8]:
        # バージョンによってキーの持ち方が違う場合があるため安全に取得
        title = item.get('title') or item.get('content', {}).get('title')
        link = item.get('link') or item.get('content', {}).get('clickThroughUrl', {}).get('url')
        
        if title and link:
            extracted_news.append({'title': title, 'link': link})

    if not extracted_news:
        print("有効なニュース項目が見つかりませんでした。")
    else:
        texts_to_translate = [news['title'] for news in extracted_news]
        
        print(f"DeepL APIで {len(texts_to_translate)} 件のタイトルを日本語に翻訳中...")
        results = translator.translate_text(texts_to_translate, target_lang="JA")
        
        news_list = []
        for news, translated in zip(extracted_news, results):
            news_list.append({
                'タイトル (日本語)': translated.text,
                '原文タイトル': news['title'],
                'リンク': f"<a href='{news['link']}' target='_blank'>記事を読む</a>"
            })
        
        df_news = pd.DataFrame(news_list)
        display(HTML(df_news.to_html(escape=False)))
        print("完了しました！")
```

```python
import os
import urllib.request
import feedparser
import pandas as pd
import deepl
from IPython.display import display, HTML

# 1. DeepLクライアントの初期化
api_key = os.environ.get("DEEPL_API_KEY")
if not api_key:
    raise ValueError("環境変数 'DEEPL_API_KEY' が設定されていません。")

translator = deepl.DeepLClient(api_key)

# 2. Google NewsのRSS（シンガポールの金融・株式市場に関するニュースを検索）
query = "Singapore stock market finance"
url = f"https://news.google.com/rss/search?q={urllib.parse.quote(query)}&hl=en-SG&gl=SG&ceid=SG:en"

req = urllib.request.Request(
    url, 
    headers={
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36'
    }
)

try:
    print("Google Newsからシンガポールの金融ニュースを取得中...")
    with urllib.request.urlopen(req, timeout=15) as response:
        xml_data = response.read()
    
    feed = feedparser.parse(xml_data)
    
    if not feed.entries:
        print("ニュースが見つかりませんでした。")
    else:
        # 上位8件を抽出
        entries_to_translate = feed.entries[:8]
        texts_to_translate = [entry.title for entry in entries_to_translate if entry.title]
        
        print(f"DeepL APIで {len(texts_to_translate)} 件のタイトルを日本語に翻訳中...")
        results = translator.translate_text(texts_to_translate, target_lang="JA")
        
        news_list = []
        for entry, translated in zip(entries_to_translate, results):
            news_list.append({
                'タイトル (日本語)': translated.text,
                '原文タイトル': entry.title,
                '公開日時': entry.get('published', 'N/A'),
                'リンク': f"<a href='{entry.link}' target='_blank'>記事を読む</a>"
            })
        
        df_news = pd.DataFrame(news_list)
        display(HTML(df_news.to_html(escape=False, render_links=True)))
        print("完了しました！")

except Exception as e:
    print(f"エラーが発生しました: {e}")
```
